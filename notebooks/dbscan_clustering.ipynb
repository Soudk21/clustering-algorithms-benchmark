```python
# ======================================================
# 1. Imports and global settings
# ======================================================
import numpy as np
import pandas as pd
import time

from sklearn.preprocessing import LabelEncoder
from sklearn.decomposition import PCA
from sklearn.metrics import precision_score, recall_score, f1_score

import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")  # seaborn style for all plots

# ======================================================
# 2. Load datasets, standardize features, encode labels
# ======================================================

dataset_names = [
    "banknote",
    "ionosphere",
    "sonar",
    "blobs",
    "varied",
    "wine",
    "flame",
    "glass",
    "iris"
]

data = {}

for name in dataset_names:
    # If your CSVs have headers, change header=None to header=0
    df = pd.read_csv(f"{name}.csv", header=None)

    X = df.iloc[:, :-1].values.astype(float)  # features
    y_raw = df.iloc[:, -1].values            # original labels (could be strings)

    # Standardization: zero mean, unit variance
    mean = X.mean(axis=0)
    std = X.std(axis=0)
    std[std == 0.0] = 1.0  # avoid division by zero
    X_std = (X - mean) / std

    # Encode labels into integers 0..C-1 for evaluation / np.bincount
    le = LabelEncoder()
    y_enc = le.fit_transform(y_raw)

    data[name] = {
        "X": X_std,
        "y": y_enc,
        "label_encoder": le
    }

    print(f"{name}: X shape = {X_std.shape}, unique encoded labels = {np.unique(y_enc)}")


# ======================================================
# 3. DBSCAN (from scratch, no sklearn)
# ======================================================
def dbscan_from_scratch(X, eps=0.5, min_samples=5):
    """
    Simple DBSCAN implementation from scratch.

    Parameters:
        X: array of shape (n_samples, n_features)
        eps: neighborhood radius
        min_samples: minimum number of points to form a dense region

    Returns:
        labels: array of shape (n_samples,), cluster labels
                -1 indicates noise.
    """
    X = np.asarray(X)
    n = X.shape[0]

    # Precompute pairwise distances (Euclidean)
    dists = np.linalg.norm(X[:, None, :] - X[None, :, :], axis=2)

    labels = np.full(n, -1, dtype=int)  # -1 means noise/unassigned
    visited = np.zeros(n, dtype=bool)

    cluster_id = 0

    for i in range(n):
        if visited[i]:
            continue

        visited[i] = True
        # Neighborhood of point i
        neighbors = np.where(dists[i] <= eps)[0]

        if neighbors.size < min_samples:
            # mark as noise for now; might become border point later
            labels[i] = -1
        else:
            # create new cluster
            labels[i] = cluster_id

            # seeds: points to be processed/expanded
            seeds = list(neighbors)
            if i in seeds:
                seeds.remove(i)

            seeds_mask = np.zeros(n, dtype=bool)
            seeds_mask[seeds] = True

            while seeds:
                current_point = seeds.pop()
                seeds_mask[current_point] = False

                if not visited[current_point]:
                    visited[current_point] = True
                    current_neighbors = np.where(dists[current_point] <= eps)[0]

                    if current_neighbors.size >= min_samples:
                        # add new neighbors to seeds if not already in it
                        for nb in current_neighbors:
                            if not seeds_mask[nb] and labels[nb] == -1:
                                seeds.append(nb)
                                seeds_mask[nb] = True

                # assign to cluster if it is noise or unassigned
                if labels[current_point] == -1:
                    labels[current_point] = cluster_id

            cluster_id += 1

    return labels


# ======================================================
# 4. Evaluation function (uses encoded integer labels)
# ======================================================
def evaluate_clustering_with_labels(y_true, cluster_labels):
    """
    Evaluate clustering using known labels.

    Noise points (cluster_labels == -1) are excluded from evaluation.

    Assumes y_true are encoded integers 0..C-1.

    Returns a dict with:
        purity, accuracy, precision_macro, recall_macro, f1_macro, n_eval_samples, coverage
    coverage = n_eval_samples / total_samples
    """
    y_true = np.asarray(y_true)
    cluster_labels = np.asarray(cluster_labels)

    # mask to ignore noise
    mask = cluster_labels != -1
    if mask.sum() == 0:
        # no points assigned to any cluster
        return {
            "purity": 0.0,
            "accuracy": 0.0,
            "precision_macro": 0.0,
            "recall_macro": 0.0,
            "f1_macro": 0.0,
            "n_eval_samples": 0,
            "coverage": 0.0
        }

    y_eval = y_true[mask]
    c_eval = cluster_labels[mask]

    # Map each cluster to the majority true label within that cluster
    y_pred = np.zeros_like(y_eval)

    unique_clusters = np.unique(c_eval)
    for c in unique_clusters:
        idx = (c_eval == c)
        true_labels_in_cluster = y_eval[idx]
        # majority vote (y_eval are non-negative ints)
        majority_label = np.bincount(true_labels_in_cluster).argmax()
        y_pred[idx] = majority_label

    accuracy = (y_pred == y_eval).mean()
    purity = accuracy  # with majority mapping, purity == accuracy

    precision_macro = precision_score(y_eval, y_pred, average="macro", zero_division=0)
    recall_macro = recall_score(y_eval, y_pred, average="macro", zero_division=0)
    f1_macro = f1_score(y_eval, y_pred, average="macro", zero_division=0)

    coverage = len(y_eval) / len(y_true)

    return {
        "purity": float(purity),
        "accuracy": float(accuracy),
        "precision_macro": float(precision_macro),
        "recall_macro": float(recall_macro),
        "f1_macro": float(f1_macro),
        "n_eval_samples": int(len(y_eval)),
        "coverage": float(coverage)
    }


# ======================================================
# 5. Automatic tuning: grid search over eps and min_samples
# ======================================================
def run_dbscan_grid(X_pca, y, eps_values, min_samples_values):
    """
    Runs DBSCAN (from scratch) for all combinations of eps and min_samples.
    Returns a DataFrame with metrics per configuration.
    """
    records = []

    for eps in eps_values:
        for ms in min_samples_values:
            start = time.time()
            cluster_labels = dbscan_from_scratch(X_pca, eps=eps, min_samples=ms)
            runtime = time.time() - start

            metrics = evaluate_clustering_with_labels(y, cluster_labels)
            rec = {
                "eps": eps,
                "min_samples": ms,
                "runtime_sec": runtime,
                "purity": metrics["purity"],
                "accuracy": metrics["accuracy"],
                "precision_macro": metrics["precision_macro"],
                "recall_macro": metrics["recall_macro"],
                "f1_macro": metrics["f1_macro"],
                "n_eval_samples": metrics["n_eval_samples"],
                "coverage": metrics["coverage"]
            }
            records.append(rec)

    df = pd.DataFrame(records)
    return df


def pick_best_config(df, min_coverage=0.5):
    """
    Pick best configuration from grid search:

    - Prefer configurations with coverage >= min_coverage (if any exist).
    - Among them, maximize f1_macro, then purity, then coverage.
    - If no configuration has coverage >= min_coverage, pick from all
      non-zero-eval configs using same ranking.
    """
    df_valid = df[df["n_eval_samples"] > 0].copy()
    if df_valid.empty:
        # No non-noise clustering; just return the row with largest coverage
        return df.sort_values(by=["coverage", "f1_macro", "purity"],
                              ascending=False).iloc[0]

    df_cov = df_valid[df_valid["coverage"] >= min_coverage]
    if not df_cov.empty:
        df_to_search = df_cov
    else:
        df_to_search = df_valid

    df_sorted = df_to_search.sort_values(
        by=["f1_macro", "purity", "coverage"],
        ascending=False
    ).reset_index(drop=True)

    return df_sorted.iloc[0]


# ======================================================
# 6. Main loop: PCA, tuning, best config, plots, summary table
# ======================================================

# Grid for automatic tuning
eps_values = np.linspace(0.2, 2.0, 10)    # from 0.2 to 2.0
min_samples_values = [3, 5, 8, 10]

best_results = []  # will collect best config per dataset

for name in dataset_names:
    print("\n" + "=" * 60)
    print(f"Dataset: {name}")
    print("=" * 60)

    X = data[name]["X"]
    y = data[name]["y"]
    le = data[name]["label_encoder"]

    # PCA to 2D
    pca = PCA(n_components=2)
    X_pca = pca.fit_transform(X)
    evr = pca.explained_variance_ratio_
    print(f"PCA explained variance ratio (PC1, PC2): {evr}")

    # Grid search for DBSCAN parameters
    grid_df = run_dbscan_grid(X_pca, y, eps_values, min_samples_values)

    # Select best configuration (automatic tuning)
    best = pick_best_config(grid_df, min_coverage=0.5)
    best_eps = float(best["eps"])
    best_ms = int(best["min_samples"])

    print("\nBest configuration (automatic tuning):")
    print(best)

    # Re-run DBSCAN with best hyperparameters to get final clustering
    best_cluster_labels = dbscan_from_scratch(X_pca, eps=best_eps, min_samples=best_ms)
    final_metrics = evaluate_clustering_with_labels(y, best_cluster_labels)

    # Store summary for this dataset
    result_row = {
        "dataset": name,
        "eps": best_eps,
        "min_samples": best_ms,
        "purity": final_metrics["purity"],
        "accuracy": final_metrics["accuracy"],
        "precision_macro": final_metrics["precision_macro"],
        "recall_macro": final_metrics["recall_macro"],
        "f1_macro": final_metrics["f1_macro"],
        "n_eval_samples": final_metrics["n_eval_samples"],
        "coverage": final_metrics["coverage"]
    }
    best_results.append(result_row)

    # -----------------------------
    # Plots for this dataset
    # -----------------------------

    # (a) PCA scatter – true labels
    plt.figure(figsize=(6, 5))
    scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y, s=30)
    plt.xlabel("PC1")
    plt.ylabel("PC2")
    plt.title(f"{name} – PCA colored by true labels")

    unique_y = np.unique(y)
    handles = []
    for lab in unique_y:
        # Decode label for legend
        original_label = le.inverse_transform([lab])[0]
        handles.append(
            plt.Line2D([], [], marker="o", linestyle="",
                       label=str(original_label))
        )
    plt.legend(handles=handles, title="True labels",
               bbox_to_anchor=(1.05, 1), loc="upper left")
    plt.tight_layout()
    plt.show()

    # (b) PCA scatter – DBSCAN clusters
    plt.figure(figsize=(6, 5))
    unique_clusters = np.unique(best_cluster_labels)

    for c in unique_clusters:
        mask = (best_cluster_labels == c)
        if c == -1:
            plt.scatter(X_pca[mask, 0], X_pca[mask, 1],
                        marker="x", s=40, label="Noise")
        else:
            plt.scatter(X_pca[mask, 0], X_pca[mask, 1],
                        s=30, label=f"Cluster {c}")

    plt.xlabel("PC1")
    plt.ylabel("PC2")
    plt.title(f"{name} – PCA colored by DBSCAN clusters\n"
              f"(eps={best_eps:.2f}, min_samples={best_ms})")
    plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
    plt.tight_layout()
    plt.show()

    # (c) Bar chart of metrics for best configuration
    metric_names = ["purity", "accuracy", "precision_macro",
                    "recall_macro", "f1_macro"]
    metric_display_names = [
        "Purity",
        "Accuracy",
        "Precision (macro)",
        "Recall (macro)",
        "F1 (macro)"
    ]
    metric_values = [final_metrics[m] for m in metric_names]

    plt.figure(figsize=(7, 4))
    plt.bar(metric_display_names, metric_values)
    plt.ylim(0.0, 1.05)
    plt.ylabel("Score")
    plt.title(f"{name} – DBSCAN performance (PCA, best tuned)")
    plt.xticks(rotation=30, ha="right")
    plt.tight_layout()
    plt.show()


# ======================================================
# 7. Summary table: best tuning & evaluation per dataset
# ======================================================
best_results_df = pd.DataFrame(best_results)

print("\n\n=== Summary: Best automatic tuning & clustering evaluation per dataset ===")
print(best_results_df)

# If you want to see it as a nice table in Colab, just evaluate 'best_results_df'
best_results_df
```
